# Colab Template Launcher — Universal Runtime Bridge

Run this single cell once (`Runtime -> Run all`) to connect this Colab VM to your Android app.
Once running, all templates, task controls, library updates, and runtime shutdown are controlled directly from the Android app.

```python
# Single-cell setup & persistent runtime daemon
import os
import subprocess
import sys
from google.colab import drive

REPO_OWNER = 'Bhavin-Viramgama'
REPO_URL = f'https://github.com/{REPO_OWNER}/colab-template-library.git'
REPO_DIR = '/content/colab-template-library'

drive.mount('/content/drive')

if os.path.exists(REPO_DIR):
    subprocess.run(['git', '-C', REPO_DIR, 'pull', '--ff-only'], check=False)
else:
    subprocess.run(['git', 'clone', REPO_URL, REPO_DIR], check=True)

subprocess.run(['pip', 'install', '-q', '-r', os.path.join(REPO_DIR, 'runtime', 'requirements.txt')], check=False)
subprocess.run(['apt-get', 'install', '-y', '-qq', 'aria2'], capture_output=True, check=False)

sys.path.insert(0, REPO_DIR)
from runtime.colab_daemon import start_daemon

start_daemon(repo_dir=REPO_DIR)
```
